```python
%pip install opencv-python pillow matplotlib -q
import cv2
import matplotlib.pyplot as plt
from PIL import Image
from pathlib import Path
img = cv2.imread(str(Path.home() / "Downloads" / "image.jpg"))
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
images = [
    rgb,
    cv2.cvtColor(img, cv2.COLOR_BGR2GRAY),
    cv2.cvtColor(img, cv2.COLOR_BGR2HSV),
    Image.fromarray(rgb).convert("CMYK")
]
for i, (name, image) in enumerate(zip(["Original", "Gray", "HSV", "CMYK"], images)):
    plt.subplot(1, 4, i+1)
    plt.imshow(image, cmap="gray" if i == 1 else None)
    plt.title(name)
    plt.axis("off")
plt.show()
```

```python
%pip install opencv-python matplotlib -q
import cv2
import matplotlib.pyplot as plt
from pathlib import Path
img = cv2.imread(str(Path.home() / "Downloads" / "image.jpg"), 0)
eq = cv2.equalizeHist(img)
plt.subplot(2,2,1)
plt.imshow(img, cmap="gray")
plt.title("Original")
plt.subplot(2,2,2)
plt.imshow(eq, cmap="gray")
plt.title("Equalized")
plt.subplot(2,2,3)
plt.hist(img.ravel(), 256, [0,256])
plt.title("Original Histogram")
plt.subplot(2,2,4)
plt.hist(eq.ravel(), 256, [0,256])
plt.title("Equalized Histogram")
plt.tight_layout()
plt.show()
```

```python
%pip install opencv-python numpy matplotlib -q
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
img = cv2.imread(str(Path.home() / "Downloads" / "image.jpg"), 0)
_, img = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
k = np.ones((5,5), np.uint8)
images = [
    img,
    cv2.dilate(img, k),
    cv2.erode(img, k),
    cv2.morphologyEx(img, cv2.MORPH_OPEN, k),
    cv2.morphologyEx(img, cv2.MORPH_CLOSE, k)]
names = ["Original", "Dilation", "Erosion", "Opening", "Closing"]
for i in range(5):
    plt.subplot(1,5,i+1)
    plt.imshow(images[i], cmap="gray")
    plt.title(names[i])
    plt.axis("off")
plt.show()
```

```python
%pip install opencv-python numpy matplotlib -q
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
img = cv2.imread(str(Path.home() / "Downloads" / "image.jpg"), 0)
f = np.fft.fftshift(np.fft.fft2(img))
r, c = img.shape
y, x = np.ogrid[:r, :c]
low = (x-c//2)**2 + (y-r//2)**2 <= 30**2
high = ~low
lp = np.abs(np.fft.ifft2(np.fft.ifftshift(f*low)))
hp = np.abs(np.fft.ifft2(np.fft.ifftshift(f*high)))
images = [img, np.log1p(abs(f)), lp, hp]
names = ["Original", "Fourier", "Low Pass", "High Pass"]
for i in range(4):
    plt.subplot(1,4,i+1)
    plt.imshow(images[i], cmap="gray")
    plt.title(names[i])
    plt.axis("off")
plt.show()
```
